```python
from sklearn.linear_model import LinearRegression

def celsius_to_fahrenheit(c):
    return c * 1.8 + 32

celsius = np.array([-40,-10,0,10,25,37,100]).reshape(-1,1) # X features
fahrenheit = np.array([-40,14,32,50,77,98.6,212]) # Y labels

model = LinearRegression()
model.fit(celsius, fahrenheit)
print(round(model.coef_[0], 4))
print(round(model.intercept_, 4))

print(round(model.predict([[-40]])[0], 2))
print(celsius_to_fahrenheit(-40))

```

```python
import time

size = 1_000_000

py_list = list(range(size))
np_arr = np.arange(size)

start = time.perf_counter()
result_list = [x * 2 for x in py_list]
t_list = time.perf_counter() - start

start = time.perf_counter()
result_np = np_arr * 2
t_np = time.perf_counter() - start

print(f"Python List: {t_list:.4f} sec")
print(f"Numpy array: {t_np:.4f} sec")
print(f"Numpy is {t_list/max(t_np, 1e-9):.0f}x faster")
```

```python
rng = np.random.default_rng(7)
revenue = rng.integers(50,200, size=(5,12))
hotels = np.array(["Hotel A", "Hotel B", "Hotel C", "Hotel D", "Hotel E"])
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

print("revenue matrix shape:", revenue.shape, "\n")

total_per_hotel = revenue.sum(axis=1)
best_hotel = hotels[total_per_hotel.argmax()]

print("Total revenue per hotel: ", dict(zip(hotels, total_per_hotel)))
print("Best hotel of the year: ", best_hotel)

best_month_idx = revenue.argmax(axis=1)
print("Best month per hotel: ", dict(zip(hotels, months[best_month_idx])))

growth = (revenue[:, 1:] - revenue[:, :-1]) / revenue[:, :-1] * 100
print("\n Growth matrix shape:", growth.shape, "(5 hotels x 11 months to months changes)")

drops = growth < -30
print("Number of drops larger than 30%:", drops.sum())
for h,m in zip(*np.where(drops)):
    print(f"{hotels[h]} fell {abs(growth[h,m]):.0f}% going from {months[m]} to {months[m+1]}")
```
